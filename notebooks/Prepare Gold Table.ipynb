```python
df_sales = spark.read.table('sales_forecast_project.silver.sales')
df_products = spark.read.table('sales_forecast_project.silver.products_data')
```

```python
df_sales.createOrReplaceTempView("df_sales")
df_products.createOrReplaceTempView("df_products")
```

```python
%sql
CREATE OR REPLACE TEMP VIEW sales_merge_product_dim 
AS
select * EXCEPT (ingestion_time) from df_sales s
LEFT JOIN (select * EXCEPT (ingestion_time) from df_products) p
ON s.SKU = p.Product_ID;
```

#### Sales Qty By Region

```python
%sql
SELECT Region, SUM(Sales_Quantity) AS Total_Sales
FROM sales_merge_product_dim
GROUP BY Region;
```

#### Sales Qty By Region, Brand & Category

```python
%sql
SELECT Region, Category, Brand,  SUM(Sales_Quantity) AS Total_Sales_Qty
FROM sales_merge_product_dim
GROUP BY Region, Category, Brand;
```

#### Top 10 Products

```python
%sql
SELECT Region, Category, Brand,  SUM(Sales_Quantity) AS Total_Sales_Qty
FROM sales_merge_product_dim
GROUP BY Region, Category, Brand;
```


#### Sales by Year, Month

```python
%sql
SELECT YEAR(Date) AS Year, 
MONTH(Date) AS Month,
Region,
Category,
Brand,
SUM(Sales_Quantity) AS Total_Sales_Qty
FROM sales_merge_product_dim
GROUP BY Year, Month, Region, Category, Brand;
```
