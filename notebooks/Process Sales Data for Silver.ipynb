```python
from pyspark.sql import functions as F
df = spark.read.table("sales_forecast_project.bronze.sales")
```

```python
df = df.drop("_rescued_data")
df = df.withColumn("Date",F.coalesce(F.expr("try_to_timestamp(Date, 'dd-MM-yyyy')"),
        F.expr("try_to_timestamp(Date, 'yyyy-MM-dd')"))
)
df = df.withColumn("Date", F.coalesce(F.col("Date"), F.lit("1900-01-01").cast("date")))
df = df.withColumn("Date", F.to_date("Date"))
df = df.withColumn("Sales_Quantity", F.coalesce(F.col("Sales_Quantity"), F.lit(0)))
df = df.withColumn("Region", F.coalesce(F.col("Region"), F.lit("NA")))
df = df.withColumn("SKU", F.coalesce(F.col("SKU"), F.lit("NA")))
```

```python
df.createOrReplaceTempView("sales_temp")
```

```python
df.write.mode("overwrite").saveAsTable("sales_forecast_project.silver.sales")
```
